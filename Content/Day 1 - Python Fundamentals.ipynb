  <h4 style="color: #312E81; margin: 0 0 8px 0; font-weight: 600; font-size: 1.1em; display: flex; align-items: center; gap: 6px;">
    🔹 Low-Level vs. High-Level Programming Languages
  </h4>
  <div style="color: #334155; line-height: 1.6; font-size: 0.95em;">
    Every programming language in history sits somewhere on a <strong>level spectrum</strong> — from raw machine code at the bottom to elegant, human-readable code at the top.<br><br>
    <strong>🔻 Low-Level Languages</strong> are languages that communicate directly with the hardware. They give programmers complete control over memory and CPU registers, but are extremely difficult to read and write:<ul>
      <li><strong>Machine Code (Level 0 — Binary)</strong>: The only language a CPU truly understands. Everything is pure 0s and 1s. Example: <code>01001000 10001001 11000011</code>. No human writes this directly — it is generated automatically.</li>
      <li><strong>Assembly Language (Level 1)</strong>: A tiny step above binary. Uses short human-readable mnemonics like <code>MOV AX, 5</code> or <code>ADD AX, BX</code> that map almost 1:1 to binary. Still extremely tedious and CPU-specific. Used in OS kernels and embedded chips.</li>
      <li><strong>C Language (Level 2)</strong>: Often considered the lowest "high-level" language. Requires manual memory management (<code>malloc</code>, <code>free</code>). Powers operating system kernels like Linux, Windows, and macOS. Very fast but error-prone for beginners.</li>
    </ul>
    <strong>🔺 High-Level Languages</strong> are designed to be readable by humans. They hide all hardware complexity behind friendly syntax:<ul>
      <li><strong>C++ / Java (Level 3)</strong>: Powerful object-oriented languages. Java compiles to "bytecode" that runs on a virtual machine. Still requires explicit type declarations and more boilerplate code.</li>
      <li><strong>Python / JavaScript / Ruby (Level 4 — Very High Level)</strong>: The most human-readable and beginner-friendly tier. Python reads almost like English — no type declarations needed, no manual memory management. The interpreter or engine handles all hardware translation automatically.</li>
    </ul>
    <strong>📌 Where does Python sit?</strong> Python is a <strong>very high-level, general-purpose language</strong>. It is so abstracted from hardware that you never think about RAM allocation or CPU registers — you just describe *what* you want, and Python figures out *how* to tell the machine.
  </div>
</div>


<div style="padding: 16px; border-left: 4px solid #4F46E5; background-color: #EEF2F6; border-radius: 6px; margin-top: 15px; margin-bottom: 15px;">
  <h4 style="color: #312E81; margin: 0 0 8px 0; font-weight: 600; font-size: 1.1em; display: flex; align-items: center; gap: 6px;">
    🐍 What Exactly is Python? — A Deep Dive
  </h4>
  <div style="color: #334155; line-height: 1.6; font-size: 0.95em;">
    <strong>Python</strong> is a <em>free, open-source, general-purpose, high-level, interpreted programming language</em> created by <strong>Guido van Rossum</strong> and first released in <strong>1991</strong>. Despite the name, Python has nothing to do with the snake — it was named after the British comedy TV show <em>Monty Python's Flying Circus</em> that Guido enjoyed!<br><br>
    <strong>Core Properties of Python:</strong><ul>
      <li><strong>General-Purpose</strong>: Not designed for one specific task. You can build websites, data analysis pipelines, AI models, desktop applications, automation bots, and games — all in Python.</li>
      <li><strong>High-Level</strong>: Abstracts all hardware complexity. You never deal with memory addresses, CPU registers, or binary instructions.</li>
      <li><strong>Interpreted</strong>: Runs code line-by-line in real time via the Python Interpreter. No separate compilation step needed before running.</li>
      <li><strong>Dynamically Typed</strong>: You never need to declare the "type" of a variable in advance. Python figures it out automatically at runtime (<code>x = 5</code> vs Java's <code>int x = 5;</code>).</li>
      <li><strong>Open Source</strong>: Python is completely free to download, use, and distribute — even in commercial products. The source code is publicly maintained at <em>python.org</em>.</li>
      <li><strong>Cross-Platform</strong>: The same Python code runs on Windows, macOS, Linux, and Raspberry Pi without modification.</li>
    </ul>
    <strong>Python Version History:</strong><ul>
      <li><strong>Python 1.x (1994)</strong>: Original release. Very limited features.</li>
      <li><strong>Python 2.x (2000–2020)</strong>: Widely adopted for a decade. Officially retired in January 2020 — no longer receives security updates.</li>
      <li><strong>Python 3.x (2008–Present)</strong>: The current, actively maintained version. Everything we write in this workshop uses Python 3. Always install Python 3!</li>
    </ul>
    <strong>Python's Design Philosophy — "The Zen of Python":</strong><br>Python was intentionally designed with a philosophy: <em>"There should be one obvious way to do it."</em> Python code should be beautiful, explicit, readable, and simple. This makes Python the most beginner-friendly professional language in the world.
  </div>
</div>


<div style="padding: 16px; border-left: 4px solid #4F46E5; background-color: #EEF2F6; border-radius: 6px; margin-top: 15px; margin-bottom: 15px;">
  <h4 style="color: #312E81; margin: 0 0 8px 0; font-weight: 600; font-size: 1.1em; display: flex; align-items: center; gap: 6px;">
    ⚙️ Interpreted vs. Compiled Languages — The Full Picture
  </h4>
  <div style="color: #334155; line-height: 1.6; font-size: 0.95em;">
    When you write code, the computer cannot run it directly — it must first be <strong>translated into machine code (binary)</strong>. There are two fundamentally different translation strategies:<br><br>
    <strong>📦 Compiled Languages (e.g. C, C++, Go, Rust)</strong><ul>
      <li>A program called a <strong>Compiler</strong> reads your <em>entire source code file</em> upfront and translates it all at once into a binary <code>.exe</code> or <code>.out</code> executable file.</li>
      <li>Once compiled, the executable runs extremely fast — the machine code is already ready.</li>
      <li>If your code has a syntax error anywhere, the compiler refuses to produce an executable and reports all errors in one go.</li>
      <li><strong>Book Analogy</strong>: Like translating an entire book from Nepali to English <em>before</em> anyone reads it. Once translated, reading is instant — but translation takes time upfront.</li>
    </ul>
    <strong>▶️ Interpreted Languages (e.g. Python, JavaScript, Ruby)</strong><ul>
      <li>A program called the <strong>Interpreter</strong> reads and executes your code <em>one line at a time</em>, in real time — no compilation step needed.</li>
      <li>If line 3 has an error but lines 1–2 are correct, lines 1–2 will run and print before the error is reported.</li>
      <li>Slightly slower than compiled code at runtime, but vastly faster for development — just save and run!</li>
      <li><strong>Translator Analogy</strong>: Like a live United Nations interpreter translating a speech <em>sentence-by-sentence</em> as it is being spoken. Immediate and interactive, but the translation happens every single time.</li>
    </ul>

🔹 **Code Example 3: Types of Comments — Single-line, Inline, and Block**
Python supports three comment styles. Single-line comments start with `#`. Inline comments sit on the same line as code. Block comments are stacked `#` lines that explain a whole section of logic above it.

```python
# ── Single-line comment ──────────────────────────────────────────
# This entire line is a comment. Python ignores it completely.
print("Single-line comment above this line.")

# ── Inline comment ───────────────────────────────────────────────
discount_rate = 0.10   # 10% discount for senior citizens
vat_rate      = 0.13   # Nepal government VAT: 13%
print(f"Discount rate: {discount_rate}, VAT rate: {vat_rate}")

# ── Block comment (stacked # lines) ──────────────────────────────
# INVOICE CALCULATION LOGIC
# Base price is fetched from the product database.
# We apply the discount first, then add VAT on the discounted price.
base_price     = 5000
discounted     = base_price * (1 - discount_rate)
final_price    = discounted * (1 + vat_rate)
print(f"Final invoice price: NPR {final_price:.2f}")
```

```python
"""
workshop_calculator.py
Python Workshop — Day 1 Demo File
Author  : Pathshala Academy
Purpose : Demonstrates docstring placement and inline comments.
Version : 1.0
"""

def calculate_area(length, width):
    """Return the area of a rectangle given its length and width."""
    return length * width   # simple multiplication

# Call the function and display result
area = calculate_area(12, 7)
print(f"Rectangle area: {area} sq units")

# We can read the docstring programmatically:
print("\nFunction documentation:")
print(calculate_area.__doc__)
```

🔹 **Code Example 5: Bitwise Operators — Working at the Binary Level**
Bitwise operators act directly on the **binary (bit-level)** representation of integers. They are used in low-level programming, hardware flags, and encryption. For reference: `6` in binary is `110` and `3` in binary is `011`.

```python
#  6  in binary = 1 1 0
#  3  in binary = 0 1 1

print('===== BITWISE OPERATORS =====')
print(f'AND  (6 & 3)  : {6 & 3}')   # 110 & 011 = 010 = 2
print(f'OR   (6 | 3)  : {6 | 3}')   # 110 | 011 = 111 = 7
print(f'XOR  (6 ^ 3)  : {6 ^ 3}')   # 110 ^ 011 = 101 = 5
print(f'NOT  (~3)     : {~3}')        # -(3+1)    =    -4
print(f'Left  Shift (3 << 2): {3 << 2}')  # 011 → 01100 = 12
print(f'Right Shift (6 >> 2): {6 >> 2}')  # 110 → 001   =  1
```

🔹 **Code Example 6: Identity Operators — `is` and `is not`**
Identity operators check whether two variables point to the **exact same object in memory** — not just whether their values are equal. Use `==` to compare *values*; use `is` to compare *object identity* (memory address).

```python
x = ['apple', 'banana']
y = ['apple', 'banana']   # same content, DIFFERENT object in memory
z = x                     # z is an alias — points to the SAME object as x

print('===== IDENTITY OPERATORS =====')
print(f'x is y     → {x is y}')      # False: two separate list objects
print(f'x is z     → {x is z}')      # True:  z is the same object as x
print(f'x is not y → {x is not y}')  # True

# Key distinction: value equality vs. object identity
print(f'\nValue check    (x == y): {x == y}')  # True  — same content
print(f'Identity check (x is y): {x is y}')   # False — different objects
```

🔹 **Code Example 7: Membership Operators — `in` and `not in`**
Membership operators test whether a value **exists inside** a collection (list, string, tuple, set, or dictionary keys). They return a Boolean `True` or `False`.

```python
fruits = ['apple', 'banana', 'mango', 'kiwi']

print('===== MEMBERSHIP OPERATORS =====')
print(f"'banana' in  fruits → {'banana' in fruits}")      # True
print(f"'grape'  in  fruits → {'grape'  in fruits}")      # False
print(f"'grape'  not in fruits → {'grape' not in fruits}")# True

# Works on strings too
course = 'Python Fundamentals Workshop'
print(f"\n'Python' in course → {'Python' in course}")     # True
print(f"'Java'   in course → {'Java'   in course}")      # False

# Works on dictionary keys
student = {'name': 'Ram', 'age': 20, 'grade': 'A'}
print(f"\n'name'  in student → {'name'  in student}")    # True
print(f"'email' in student → {'email' in student}")     # False
```

### 🎓 Practice Exercise 3: Simple Arithmetic Calculator
**Objective**: Build a simple arithmetic console calculator. The program must ask the user for two numbers, and then calculate and display their **Addition**, **Subtraction**, **Multiplication**, and **Standard Division** in a beautiful visual ledger dashboard.
📊 **Teacher's Blueprint Hint**:
1. Capture two numeric inputs and cast them immediately to floats.
2. Perform all four basic calculations: `+`, `-`, `*`, and `/`.
3. Output a beautiful text-based dashboard displaying all four answers. For division, use precision formatting `:.2f` to round the quotient cleanly!

```python
# 1. Capture user inputs
a = float(input("Enter first number (a): "))
b = float(input("Enter second number (b): "))

# 2. Perform arithmetic
addition = a + b
subtraction = a - b
multiplication = a * b
division = a / b if b != 0 else 0

# 3. Output beautiful dashboard
print("========== ARITHMETIC DASHBOARD ==========")
print(f"Input A:        {a}")
print(f"Input B:        {b}")
print("------------------------------------------")
print(f"Addition (a+b): {addition}")
print(f"Subtraction:    {subtraction}")
print(f"Multiplication: {multiplication}")
print(f"Division (a/b): {division:.2f}" if b != 0 else "Division:       ERROR (Cannot divide by 0)")
print("==========================================")
```

### 🎓 Practice Exercise 4: Temperature Conversion Program (Celsius to Fahrenheit)
**Objective**: Create an automated utility that asks the user for a temperature in **Celsius**, performs the conversion to **Fahrenheit** using the standard scientific formula: $$F = (C \times \frac{9}{5}) + 32$$, and prints a personalized f-string message showing the conversion rounded to exactly 2 decimal places.
📊 **Teacher's Blueprint Hint**:
1. Capture the Celsius input and convert it to a float.
2. Apply the mathematical formula: `fahrenheit = (celsius * 9 / 5) + 32`.
3. Display the final result using an f-string with precision formatting: `f"... {fahrenheit:.2f}°F"`.

```python
# 1. Gather Celsius temperature
celsius = float(input("Enter temperature in Celsius (°C): "))

# 2. Convert to Fahrenheit
fahrenheit = (celsius * 9 / 5) + 32

# 3. Print rounded conversion results
print(f"🌡️ Temperature Conversion Log:")
print(f"{celsius:.1f}°C is mathematically equivalent to: {fahrenheit:.2f}°F")
```
