🔹 **Code Example 1 — Simplest Function Definition & Call**

````python
# Define the function
def greet():
    print("Hello Student")

# Call the function to execute its code
greet()
````

🔹 **Code Example 2 — Function with Parameters**

````python
# Define function with a 'name' parameter
def greet_user(name):
    print(f"Hello {name}")

# Call the function, passing a string argument
greet_user("Ram")
````

🔹 **Code Example 3 — Real-World Use Case (Calculated Return Values)**

````python
# Define a mathematical helper function that returns a value
def add(a, b):
    return a + b

# Call function and store return value in a variable
result = add(5, 3)
print("Result of function calculation:", result)
````

🔹 **Code Example 1 — List Definition & Display**

````python
# Create an ordered list of strings
fruits = ["Apple", "Banana", "Mango"]
print(fruits)
````

🔹 **Code Example 2 — Appending & Mutating Elements**

````python
numbers = [10, 20, 30]

# Add 40 to the end of the list
numbers.append(40)
print("After append:", numbers)

# Mutate/modify the first element (index 0)
numbers[0] = 99
print("After modification:", numbers)
````

🔹 **Code Example 3 — Real-World Use Case (Looping Through Lists)**

````python
# List of students
students = ["Ram", "Sita", "Hari"]

# Iterate and print welcome banners
for student in students:
    print(f"Welcome to Day 3, {student}!")
````

````python
# Create an immutable tuple
colors = ("Red", "Blue", "Green")
print(colors)

# Accessing element works just like lists
print("First color:", colors[0])
````

🔹 **Code Example 1 — Dictionary Definition & Lookup**

````python
# Create a dictionary containing student details
student = {
    "name": "Ram",
    "age": 20
}

# Lookup values using key label
print("Student Name:", student["name"])
````

🔹 **Code Example 2 — Updating & Adding Fields**

````python
student = {
    "name": "Sita",
    "marks": 90
}

# Update existing value
student["marks"] = 95

# Add a completely new key-value pair
student["city"] = "Kathmandu"

print("Updated student record:", student)
````

Let's combine Lists, Dictionaries, User Inputs, Loops, and Conditionals to build a terminal database program! This script runs continuously, allowing users to add students, view records, or exit.
⚠️ **Note on Running Interactive Loops in Notebooks**:
When running this cell, it will prompt for input continuously. Click **3** to safely exit the program, which will release the execution cell!

````python
# Empty database storage list
students = []

while True:
    print("\n--- STUDENT RECORDS MENU ---")
    print("1. Add Student")
    print("2. View Students")
    print("3. Exit")
    
    choice = input("Enter choice (1-3): ")
    
    if choice == "1":
        name = input("Enter student name: ")
        age = input("Enter student age: ")
        marks = input("Enter student marks: ")
        
        # Package details into a structured dictionary
        student = {
            "name": name,
            "age": age,
            "marks": marks
        }
        
        # Save record to list database
        students.append(student)
        print(f"🎉 Student {name} Added successfully!")
        
    elif choice == "2":
        if not students:
            print("📂 Database is currently empty.")
        else:
            print(f"\nTotal Records: {len(students)}")
            for idx, student in enumerate(students, 1):
                print(f"{idx}. Name: {student['name']}, Age: {student['age']}, Marks: {student['marks']}")
                
    elif choice == "3":
        print("👋 Exiting database system. Goodbye!")
        break
        
    else:
        print("❌ Invalid option selected. Try again!")
````

We build a fully modular calculator where mathematical operations are isolated inside robust functions, operated inside an interactive loop.

````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "❌ Error: Cannot divide by zero!"
    return a / b

while True:
    print("\n--- MODULAR CALCULATOR MENU ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")
    
    choice = input("Enter choice (1-5): ")
    
    if choice == "5":
        print("👋 Exiting Calculator. Goodbye!")
        break
        
    if choice in ["1", "2", "3", "4"]:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        
        if choice == "1":
            print("Result:", add(num1, num2))
        elif choice == "2":
            print("Result:", subtract(num1, num2))
        elif choice == "3":
            print("Result:", multiply(num1, num2))
        elif choice == "4":
            print("Result:", divide(num1, num2))
    else:
        print("❌ Invalid Option selected!")
````
