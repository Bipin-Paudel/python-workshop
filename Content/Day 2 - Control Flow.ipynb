In Python, we use:<ul><li>`if`: Executes a code block if the condition is True.</li><li>`elif` (else if): Checks another condition if the previous ones were False.</li><li>`else`: Executes a fallback block of code if all prior conditions were False.</li></ul>

🔹 **Code Example 1 — Most Basic Single Condition**

````python
# Store age
age = 18

# Check condition
if age >= 18:
    # Indented block that runs if condition is met
    print("You can vote")
````

🔹 **Code Example 2 — Decisive If-Else Tree**

````python
# Store student marks
marks = 40

# Decide pass or fail status
if marks >= 40:
    print("Pass")
else:
    print("Fail")
````

🔹 **Code Example 3 — Real-World Use Case (Login Verification)**

````python
# Prompt user for access password
password = input("Enter password: ")

# Verify against a hardcoded credential
if password == "python123":
    print("Login Successful")
else:
    print("Wrong Password")
````

🔹 **Code Example 4 — Complete If-Elif-Else Decision Tree**
Checks multiple thresholds sequentially to assign a school grade status.

````python
# Set student score
score = 75

# Decide grade based on multi-stage logic boundaries
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: D")
````

For example, `range(1, 6)` generates `1, 2, 3, 4, 5`.</li><li>`while` loops MUST include a way to update variables inside the loop (e.g. `count += 1`), otherwise the loop condition will stay True forever, causing an **Infinite Loop**!</li></ul>

🔹 **Code Example 1 — The Standard For Loop**

````python
# Print numbers from 1 to 5 using a for loop
for number in range(1, 6):
    print(number)
````

🔹 **Code Example 2 — The While Loop Counter**

````python
# Start a counter variable
count = 1

# Loop while count is less or equal to 5
while count <= 5:
    print(count)
    
    # CRITICAL: increment counter to eventually stop the loop!
    count += 1
````

🔹 **Code Example 3 — Real-World Use Case (Looping Through Lists)**

````python
# Defining an attendance ledger
students = ["Ram", "Sita", "Hari"]

# Iterating through the sequence and printing each attendee name
for student in students:
    print(student)
````

Python provides three keywords to do this:<ul><li>`break`: Instantly terminates and exits the entire loop.</li><li>`continue`: Skips the rest of the current iteration and jumps straight to the next round of the loop.</li><li>`pass`: A null statement that acts as a placeholder doing absolutely nothing.</li></ul>

For example, if you are searching a large database list for a specific name, once you find the name, there is no need to check the remaining million rows. You call `break` to save computer resources! Or you can use `continue` to skip invalid email addresses in a mailing program.
  </div>

🔹 **Code Example 1 — Breaking a Loop Early**

````python
# Loop from 1 to 5
for number in range(1, 6):
    # Stop execution when number reaches 3
    if number == 3:
        break
    print(number)
````

🔹 **Code Example 2 — Skipping Iterations with Continue**

````python
# Loop from 1 to 5
for number in range(1, 6):
    # Skip number 3 completely
    if number == 3:
        continue
    print(number)
````

### Exercise 1: Even or Odd Checker
Write a program that takes an integer input from the user and checks if the number is Even or Odd using the Modulo (`%`) operator.

````python
# Ask user for input
number = int(input("Enter a number: "))

# Modulo 2 of an even number is always 0
if number % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
````

### Exercise 2: Multiplication Table Generator
Prompt the user to enter a number and generate a clean multiplication table from 1 to 10 for that number.

````python
# Prompt user for table integer
number = int(input("Enter number: "))

# Loop from 1 to 10
for i in range(1, 11):
    # Calculate product and display inline
    print(f"{number} x {i} = {number * i}")
````

### Exercise 3: Factorial Calculator
Calculate the factorial of a user-entered number (e.g. Factorial of `5` is `5 * 4 * 3 * 2 * 1 = 120`).

````python
# Prompt user for integer
number = int(input("Enter number: "))

# Variable to accumulate product
factorial = 1

# Multiply numbers sequentially from 1 to number
for i in range(1, number + 1):
    factorial *= i

print("Factorial is:", factorial)
````

### Activity 1: Number Guessing Game
We generate a random secret number between 1 and 10 and prompt the user to guess it in a single try!

````python
import random

# Generate random number between 1 and 10
secret_number = random.randint(1, 10)

# Take user guess
guess = int(input("Guess number from 1 to 10: "))

# Match guess against secret
if guess == secret_number:
    print("🎉 Correct Guess!")
else:
    print(f"❌ Wrong Guess! The number was {secret_number}")
````

### Activity 2: Star Pattern Generator
Generate a beautiful, ascending star triangle pattern using nested iteration arithmetic.

````python
# Iterating to build pattern rows
for i in range(1, 6):
    # Multiplication of string repeats character 'i' times
    print("*" * i)
````
